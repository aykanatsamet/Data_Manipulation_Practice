```python
import pandas as pd

rfm = pd.read_csv("rfm_clustered.csv")
cluster_summary = pd.read_csv("cluster_summary.csv", index_col=0)
```

```python
cluster_summary = rfm.groupby("Cluster").agg({
    "Recency": "mean",
    "Frequency": "mean",
    "Monetary": "mean",
    "CustomerID": "count"
}).rename(columns={"CustomerID": "CustomerCount"})

cluster_summary
```

```python
cluster_names = {
    1: "High-Value Loyal Customers",
    0: "Recent Occasional Buyers",
    3: "At-Risk Customers",
    2: "Low-Value Inactive Customers"
}

rfm["Segment"] = rfm["Cluster"].map(cluster_names)
```

```python
segment_summary = rfm.groupby("Segment").agg({
    "Recency": "mean",
    "Frequency": "mean",
    "Monetary": "mean",
    "CustomerID": "count"
}).rename(columns={"CustomerID": "CustomerCount"})

segment_summary
```

### High-Value Loyal Customers
This segment represents customers with high purchase frequency and monetary value,
indicating strong loyalty and long-term engagement.

### Recent Occasional Buyers
Customers in this group have made recent purchases but with lower frequency,
suggesting potential for growth through targeted engagement.

### At-Risk Customers
These customers have not purchased recently and show declining engagement,
indicating a risk of churn.

### Low-Value Inactive Customers
This segment includes customers with low spending and infrequent purchases,
requiring cost-efficient or automated marketing strategies.


## Business Implications

- Personalized marketing campaigns can be designed for each customer segment.
- High-value customers can be targeted with loyalty programs.
- At-risk customers may benefit from re-engagement strategies.
- Resource allocation can be optimized by focusing on the most valuable segments.


## Limitations

- The analysis is based on historical transactional data and does not account for future behavior.
- Customer preferences may change over time.
- The dataset represents a specific time period and business context.


## Ethical Considerations

- Customer segmentation may introduce bias if not regularly updated.
- Automated decision-making based on segments should be transparent and fair.
- Customer privacy must be preserved, and personal data should not be misused.
