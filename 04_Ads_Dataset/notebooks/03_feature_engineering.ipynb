# Feature Engineering


## Objective

The objective of feature engineering is to transform raw data into meaningful numerical
representations that can be effectively used by machine learning models.

This includes handling categorical variables, scaling numerical features,
and removing non-informative attributes.


```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler

url = "https://raw.githubusercontent.com/ameenmanna8824/DATASETS/main/Social_Network_Ads.csv"
data = pd.read_csv(url)
```

## Feature Selection

In this dataset, the `User ID` column serves only as an identifier and does not contain
predictive information. Therefore, it is removed from the feature set.


```python
data = data.drop(columns=["User ID"])
```

## Defining Target Variable

The target variable for this classification task is `Purchased`,
which indicates whether a user made a purchase after viewing an advertisement.


```python
X = data.drop("Purchased", axis=1)
y = data["Purchased"]
```

## Encoding Categorical Variables

Machine learning models require numerical input.
The categorical variable `Gender` is transformed using one-hot encoding.


```python
X = pd.get_dummies(X, columns=["Gender"], drop_first=True)
```

## Feature Scaling

Numerical features are scaled to ensure that all variables contribute equally
to the model training process.


```python
scaler = StandardScaler()

X[["Age", "EstimatedSalary"]] = scaler.fit_transform(
    X[["Age", "EstimatedSalary"]]
)
```

## Final Feature Set

After encoding and scaling, the dataset is fully prepared for model training.

## Saving Processed Data

Saving the processed dataset ensures reproducibility and
allows consistent usage across modeling notebooks.


```python
processed_df = X.copy()
processed_df["Purchased"] = y

processed_df.to_csv("ads_dataset_processed.csv", index=False)
```

## Conclusion

The dataset is now cleaned, encoded, and scaled, making it suitable
for machine learning model training.
